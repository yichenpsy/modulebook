# Module Blocks Viewer

Use this notebook to inspect `module_blocks.jsonl` as a table and compare extracted `raw_text` with the PDF.

```python
from pathlib import Path
import pandas as pd

study = "architektur_2021"

relative_work_path = Path("tools/pdf_import/data/work") / study
cwd = Path.cwd()

work_dir_candidates = [
    cwd / relative_work_path,
    cwd.parent / relative_work_path,
    cwd.parent.parent / relative_work_path,
    cwd.parent.parent.parent / relative_work_path,
    cwd / "../data/work" / study,
]

work_dir = next((path.resolve() for path in work_dir_candidates if path.exists()), None)

if work_dir is None:
    checked = "\n".join(str(path) for path in work_dir_candidates)
    raise FileNotFoundError(f"Could not find work dir for {study}. Current cwd: {cwd}\nChecked:\n{checked}")

module_blocks_path = work_dir / "module_blocks.jsonl"
pages_path = work_dir / "pages.jsonl"

print("cwd:", cwd)
print("work_dir:", work_dir)
print("module_blocks_path:", module_blocks_path)
print("pages_path:", pages_path)
```

```python
df = pd.read_json(module_blocks_path, lines=True)
df
```

```python
# Compact table for quick review
df[["module_number", "source_pages", "char_count", "raw_text"]]
```
